```python
import pandas as pd
import time
from google_play_scraper import reviews, Sort

# App details
APP_ID = "com.sraoss.dmrc"
TARGET_REVIEWS = 3000

all_reviews = []
seen_ids = set()
continuation_token = None

print("Starting review collection...")

while len(all_reviews) < TARGET_REVIEWS:

    batch, continuation_token = reviews(
        APP_ID,
        lang="en",
        country="in",
        sort=Sort.NEWEST,
        count=200,
        continuation_token=continuation_token
    )

    if not batch:
        print("No more reviews available.")
        break

    for review in batch:

        review_id = review["reviewId"]

        if review_id not in seen_ids:
            all_reviews.append(review)
            seen_ids.add(review_id)

            if len(all_reviews) >= TARGET_REVIEWS:
                break

    print(f"Collected {len(all_reviews)} / {TARGET_REVIEWS}")

    if continuation_token is None:
        print("Reached the end of available reviews.")
        break

    time.sleep(1)

print("Collection finished!")
```

```python

df = pd.DataFrame(all_reviews)

print("Dataset shape:", df.shape)

print("\nAvailable columns:")
print(df.columns.tolist())

df.head()
```

```python

selected_columns = [
    "reviewId",
    "content",
    "score",
    "at",
    "thumbsUpCount",
    "reviewCreatedVersion",
    "replyContent",
    "repliedAt"
]

df = df[selected_columns].copy()

df.head()
```

```python

df.to_csv(
    "delhi_metro_reviews.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Dataset successfully saved!")
```

```python
print("Total reviews:", len(df))

print("\nRating distribution:")
print(df["score"].value_counts().sort_index())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDate range:")
print(df["at"].min(), "to", df["at"].max())
```
