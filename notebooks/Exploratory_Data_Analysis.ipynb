```python
def clean_text(text):

    text = str(text).lower()

    # Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)

    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)

    # Remove special characters but preserve letters and numbers
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text


df['cleaned_content'] = df['content'].apply(clean_text)
```

```python
def clean_text(text):

    text = str(text).lower()

    # Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)

    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)

    # Expand common contractions
    text = re.sub(r"don't", "do not", text)
    text = re.sub(r"doesn't", "does not", text)
    text = re.sub(r"isn't", "is not", text)
    text = re.sub(r"can't", "cannot", text)
    text = re.sub(r"won't", "will not", text)

    # Preserve letters, numbers and whitespace
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text


df['cleaned_content'] = df['content'].apply(clean_text)
```

```python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

# Preserve important negative words
negation_words = {
    'no', 'not', 'nor', 'never',
    'cannot', 'none', 'neither'
}

stop_words = stop_words - negation_words
```

```python
def remove_stopwords(text):

    words = text.split()

    filtered_words = [
        word for word in words
        if word not in stop_words
    ]

    return ' '.join(filtered_words)


df['filtered_content'] = df['cleaned_content'].apply(remove_stopwords)
```

```python
# Compare star-rating sentiment with VADER sentiment

def rating_sentiment(score):
    if score <= 2:
        return 'Negative'
    elif score == 3:
        return 'Neutral'
    else:
        return 'Positive'

df['rating_sentiment'] = df['score'].apply(rating_sentiment)

comparison = pd.crosstab(
    df['rating_sentiment'],
    df['sentiment']
)

comparison
```
